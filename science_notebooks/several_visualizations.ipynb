```python
# =============================================================================
# 1. LINE PLOT
# =============================================================================
# When to use: Show trends over time or continuous data
# Best for: Time series, trends, continuous relationships
print("=" * 60)
print("VISUALIZATION 1: LINE PLOT")
print("=" * 60)
# Generate synthetic data: Temperature over 12 months
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
temperature = [15, 17, 20, 24, 28, 32, 35, 34, 30, 25, 20, 16]
humidity = [65, 60, 55, 50, 45, 40, 35, 38, 45, 55, 60, 65]

# Create a figure and an axes object
fig, ax = plt.subplots(figsize=(10, 6))

#line plot
ax.plot(months, temperature, marker='o', linewidth=2,
label='Temperature (°C)', color='red')
ax.plot(months, humidity, marker='s', linewidth=2,
label='Humidity (%)', color='blue')
ax.set_title('Temperature and Humidity Throughout the Year',
fontsize=16, fontweight='bold')
ax.set_xlabel('Month', fontsize=12)
ax.set_ylabel('Value', fontsize=12)
# Set y-axis limits from 0 to 80
ax.set_ylim(0, 80)
ax.legend(loc='best', fontsize=10)
ax.grid(True, alpha=0.8)
plt.show()
print("\n** Line plot created successfully! **")
print(" -> Key parameters: marker, linewidth, label, color")
```

```python
# =============================================================================
# 3. SCATTER PLOT
# =============================================================================
print("=" * 60)
print("VISUALIZATION 3: SCATTER PLOT")
print("=" * 60)
# Generate synthetic data: Study hours vs Exam scores
np.random.seed(89)
n_students = 50
study_hours = np.random.uniform(1, 10, n_students)
exam_scores = 40 + 5 * study_hours + np.random.normal(0, 5, n_students)
# Clip scores to valid range
exam_scores = np.clip(exam_scores, 0, 100)
# Create scatter plot
fig, ax = plt.subplots(figsize=(10, 6))
scatter = ax.scatter(study_hours, exam_scores,
                     c=exam_scores,
                     cmap='viridis',
                     s=100,
                     alpha=0.6,
                     edgecolors='black',
                     linewidth=1)
# Add colorbar
cbar = plt.colorbar(scatter, ax=ax)
cbar.set_label('Exam Score', fontsize=12)
# Add trend line
z = np.polyfit(study_hours, exam_scores, 1)
p = np.poly1d(z)
ax.plot(study_hours, p(study_hours),
        color='red',
        linestyle='--',
        linewidth=2,
        label=f'Trend: y={z[0]:.2f}x+{z[1]:.2f}')
ax.set_title('Relationship: Study Hours vs Exam Scores',
             fontsize=16, fontweight='bold')
ax.set_xlabel('Study Hours', fontsize=12)
ax.set_ylabel('Exam Score (%)', fontsize=12)
ax.set_xlim(0, 11)
ax.set_ylim(0, 100)
ax.legend(loc='lower right', fontsize=10)
ax.grid(True, alpha=0.8)
plt.show()
print("\n** Scatter plot created successfully! **")
print(" -> Key parameters: c (color), cmap, s (size), alpha, edgecolors")
print(" -> Bonus: Added trend line using np.polyfit()")
```

```python
# =============================================================================
# 4. HISTOGRAM
# =============================================================================
print("=" * 60)
print("VISUALIZATION 4: HISTOGRAM")
print("=" * 60)
# Generate synthetic data: Student ages in a university
np.random.seed(89)
ages = np.concatenate([
np.random.normal(20, 2, 300),
 # Main student body
np.random.normal(35, 5, 50)
 # Mature students
])

# Figure 1: Basic Histogram and Histogram with Density
fig1, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Simple histogram
ax1.hist(ages, bins=30, color='skyblue', edgecolor='black', alpha=0.7)
ax1.set_title('Basic Histogram', fontsize=14, fontweight='bold')
ax1.set_xlabel('Age (years)', fontsize=12)
ax1.set_ylabel('Frequency', fontsize=12)
ax1.set_xlim(0, 60)
ax1.grid(axis='y', alpha=0.8)

# Histogram with density
ax2.hist(ages, bins=30, density=True, color='lightgreen',
edgecolor='black', alpha=0.7, label='Histogram')
ax2.set_title('Histogram with Density', fontsize=14, fontweight='bold')
ax2.set_xlabel('Age (years)', fontsize=12)
ax2.set_ylabel('Density', fontsize=12)
ax2.set_xlim(0, 60)
ax2.legend()
ax2.grid(axis='y', alpha=0.8)
plt.tight_layout()
plt.show()

# Figure 2: Comparing Different Bin Sizes and Histogram with KDE (Seaborn)
fig2, (ax3, ax4) = plt.subplots(1, 2, figsize=(14, 6))

# Histogram with different bin sizes (all on ax3)
for bins, color, alpha in [(10, 'red', 0.3), (20, 'blue', 0.3), (40, 'green', 0.3)]:
    ax3.hist(ages, bins=bins, alpha=alpha,
    edgecolor='black', label=f'{bins} bins')
ax3.set_title('Comparing Different Bin Sizes', fontsize=14, fontweight='bold')
ax3.set_xlabel('Age (years)', fontsize=12)
ax3.set_ylabel('Frequency', fontsize=12)
ax3.set_xlim(0, 60)
ax3.legend()
ax3.grid(axis='y', alpha=0.8)

# Histogram with seaborn (includes KDE)
sns.histplot(ages, bins=30, kde=True, color='purple',
edgecolor='black', ax=ax4)
ax4.set_title('Histogram with KDE (Seaborn)', fontsize=14, fontweight='bold')
ax4.set_xlabel('Age (years)', fontsize=12)
ax4.set_ylabel('Count', fontsize=12)
ax4.set_xlim(0, 60)
ax4.grid(axis='y', alpha=0.8)
plt.tight_layout()
plt.show()

print("\n** Histograms created successfully! **")
print(" -> Key parameters: bins, density, alpha, edgecolor")
print(" -> Note: bins parameter significantly affects visualization!")
```

```python
# =============================================================================
# 5. BOX PLOT
# =============================================================================
print("=" * 60)
print("VISUALIZATION 5: BOX PLOT")
print("=" * 60)

np.random.seed(89)
class_a = np.random.normal(75, 10, 50)
class_b = np.random.normal(82, 8, 50)
class_c = np.random.normal(70, 15, 50)
class_d = np.random.normal(88, 6, 50)

class_c = np.append(class_c, [30, 35, 95, 98])

data = [class_a, class_b, class_c, class_d]
labels = ['Class A', 'Class B', 'Class C', 'Class D']

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

bp = axes[0].boxplot(data, labels=labels, patch_artist=True,
                     notch=True, showmeans=True,
                     boxprops=dict(
                         facecolor='lightblue',
                         edgecolor='black',
                         linewidth=1.5
                     ),
                     whiskerprops=dict(color='black', linewidth=1.5),
                     capprops=dict(color='black', linewidth=1.5),
                     medianprops=dict(color='red', linewidth=2),
                     meanprops=dict(marker='D', markerfacecolor='green',
                                    markeredgecolor='black', markersize=8))
axes[0].set_title('Test Scores by Class (Matplotlib)', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Class', fontsize=12)
axes[0].set_ylabel('Score (%)', fontsize=12)
axes[0].set_ylim(0, 110)
axes[0].grid(axis='y', alpha=0.3)

from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], color='red', linewidth=2, label='Median'),
    Line2D([0], [0], marker='D', color='w', markerfacecolor='green',
           markeredgecolor='black', markersize=8, label='Mean')
]
axes[0].legend(handles=legend_elements, loc='lower right')

df_scores = pd.DataFrame({
'Class': ['A']*len(class_a) +
['B']*len(class_b) +
['C']*len(class_c) +
['D']*len(class_d),
'Score': np.concatenate([class_a, class_b, class_c, class_d])
})
sns.boxplot(data=df_scores, x='Class', y='Score', ax=axes[1],
palette='Set2', linewidth=1.5)
axes[1].set_title('Test Scores by Class (Seaborn)', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Class', fontsize=12)
axes[1].set_ylabel('Score (%)', fontsize=12)
axes[1].set_ylim(0, 110)
axes[1].grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
print("\n** Box plots created successfully! **")
print(" -> Key parameters: patch_artist, notch, showmeans")
print(" -> Note: Red line = median, Green diamond = mean")
print(" -> Class C has outliers (points beyond whiskers)")
```

```python
#kernel density plot visualizes the probability function of a continuos vaibale
#bring in a continuous data representation about probability function
#by showing smooth distributions shapes
#know as pdf

print("-"*60)
print("VISUALIZATION 7 : PDF PLOT ")
print("-"*60)

#synthehic data
np.random.seed(89)
                            #(media,std,sample size)
heights_group1=np.random.normal(170,8,300)
heights_group2=np.random.normal(165,7,300)
heights_group3=np.random.normal(175,9,300)


#create the figure
# how many axes , for a single density plot
fig, (ax1,ax2)=plt.subplots(1,2,figsize=(14,6))
ax1.hist(heights_group1, bins=30, density=True, alpha=0.5,
        color='lightblue', edgecolor='black', label='Histogram')
sns.kdeplot(heights_group1, ax=ax1, color='darkblue',
            linewidth=2, label='PDF')
ax1.set_title('Density Plot: Single Distribution',
fontsize=14, fontweight='bold')
ax1.set_xlabel('Height (cm)', fontsize=12)
ax1.set_ylabel('Density', fontsize=12)
ax1.legend()
ax1.grid(alpha=0.8)

#Multiple density plots
sns.kdeplot(heights_group1, ax=ax2, color='red',
            linewidth=2, label='Group 1', fill=True, alpha=0.3)
sns.kdeplot(heights_group2, ax=ax2, color='blue',
            linewidth=2, label='Group 2', fill=True, alpha=0.3)
sns.kdeplot(heights_group3, ax=ax2, color='green',
            linewidth=2, label='Group 3', fill=True, alpha=0.3)
ax2.set_title('Comparing Multiple Distributions',
              fontsize=14, fontweight='bold')
ax2.set_xlabel('Height (cm)', fontsize=12)
ax2.set_ylabel('Density', fontsize=12)
ax2.legend()
ax2.grid(alpha=0.8)
plt.show()
```

```python
# =============================================================================
# 9. ERROR BARS
# =============================================================================
print("=" * 60)
print("VISUALIZATION 9: ERROR BARS")
print("=" * 60)

np.random.seed(42)
doses = [10, 20, 30, 40, 50]
n_samples = 30

effectiveness_means = [25, 45, 65, 75, 78]
effectiveness_stds = [5, 7, 8, 6, 9]

data_by_dose = []
for mean, std in zip(effectiveness_means, effectiveness_stds):
  data_by_dose.append(np.random.normal(mean, std, n_samples))

means = [np.mean(data) for data in data_by_dose]
stds = [np.std(data) for data in data_by_dose]
sems = [np.std(data) / np.sqrt(len(data)) for data in data_by_dose]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

ax1.errorbar(doses, means, yerr=stds,
fmt='o-', color='blue', linewidth=2, markersize=8,
capsize=5, capthick=2, ecolor='darkblue',
label='Mean ± SD')
ax1.set_title('Error Bars: Standard Deviation',
fontsize=14, fontweight='bold')
ax1.set_xlabel('Dose (mg)', fontsize=12)
ax1.set_ylabel('Effectiveness (%)', fontsize=12)
ax1.legend()
ax1.grid(alpha=0.8)
ax1.set_xlim(0, 60)
ax1.set_ylim(0, 100)

ax2.errorbar(doses, means, yerr=sems,
fmt='s-', color='red', linewidth=2, markersize=8,
capsize=5, capthick=2, ecolor='darkred',
label='Mean ± SEM')
ax2.set_title('Error Bars: Standard Error of Mean',
fontsize=14, fontweight='bold')
ax2.set_xlabel('Dose (mg)', fontsize=12)
ax2.set_ylabel('Effectiveness (%)', fontsize=12)
ax2.legend()
ax2.grid(alpha=0.8)
ax2.set_xlim(0, 60)
ax2.set_ylim(0, 100)
plt.show()
```

```python
# =============================================================================
# 13. PARALLEL SETS (ALLUVIAL/SANKEY DIAGRAM)
# =============================================================================
print("=" * 60)
print("VISUALIZATION 13: PARALLEL SETS")
print("=" * 60)
# Import plotly
import plotly.graph_objects as go
import plotly.express as px
# Generate synthetic data: Customer Journey (Device → Browser → Purchase)
np.random.seed(89)
n_customers = 500
devices = np.random.choice(['Mobile', 'Desktop', 'Tablet'],
n_customers, p=[0.5, 0.4, 0.1])
browsers = []
purchases = []
# Add some realistic associations
for device in devices:
    if device == 'Mobile':
        browsers.append(np.random.choice(['Chrome', 'Safari', 'Firefox'],
p=[0.5, 0.4, 0.1]))
    elif device == 'Desktop':
        browsers.append(np.random.choice(['Chrome', 'Safari', 'Firefox'],
p=[0.6, 0.2, 0.2]))
    else: # Tablet
        browsers.append(np.random.choice(['Chrome', 'Safari', 'Firefox'],
p=[0.3, 0.6, 0.1]))
for i, (device, browser) in enumerate(zip(devices, browsers)):
    # Purchase likelihood varies by device and browser
    if device == 'Desktop' and browser == 'Chrome':
        purchase_prob = 0.7
    elif device == 'Mobile' and browser == 'Safari':
        purchase_prob = 0.6
    else:
        purchase_prob = 0.4
    purchases.append(np.random.choice(['Purchased', 'Abandoned'],
p=[purchase_prob, 1-purchase_prob]))
# Create DataFrame
df_journey = pd.DataFrame({
'Device': devices,
'Browser': browsers,
'Outcome': purchases
})
# 1. Parallel Categories with Plotly
fig1 = px.parallel_categories(
df_journey,
dimensions=['Device', 'Browser', 'Outcome'],
color_continuous_scale=px.colors.sequential.Viridis,
title='Customer Journey: Device → Browser → Purchase Outcome')
fig1.update_layout(
width=1000,
height=600,
font=dict(size=12)
)
# Display in notebook
fig1.show()
print("\n** Parallel sets visualization created! **")
print(" -> Interactive visualization - hover over flows to see details")
# 2. Create a Sankey Diagram (alternative representation)
# Prepare data for Sankey
# Create unique labels
all_labels = (
[f"Device: {d}" for d in df_journey['Device'].unique()] +
[f"Browser: {b}" for b in df_journey['Browser'].unique()] +
[f"Outcome: {o}" for o in df_journey['Outcome'].unique()]
)
# Create label to index mapping
label_to_idx = {label: idx for idx, label in enumerate(all_labels)}
# Count flows
flows = []
# Device → Browser
device_browser = df_journey.groupby(['Device',
'Browser']).size().reset_index(name='count')
for _, row in device_browser.iterrows():
    flows.append({
'source': label_to_idx[f"Device: {row['Device']}"],
'target': label_to_idx[f"Browser: {row['Browser']}"],
'value': row['count']
})
# Browser → Outcome
browser_outcome = df_journey.groupby(['Browser',
'Outcome']).size().reset_index(name='count')
for _, row in browser_outcome.iterrows():
    flows.append({
'source': label_to_idx[f"Browser: {row['Browser']}"],
'target': label_to_idx[f"Outcome: {row['Outcome']}"],
'value': row['count']
})
# Create Sankey diagram
fig2 = go.Figure(data=[go.Sankey(
node=dict(
pad=15,
thickness=20,
line=dict(color='black', width=0.5),
label=all_labels,
color=['lightblue', 'lightblue', 'lightblue', # Devices
'lightgreen', 'lightgreen', 'lightgreen', # Browsers
'lightcoral', 'lightyellow'] # Outcomes
),
link=dict(
source=[f['source'] for f in flows],
target=[f['target'] for f in flows],
value=[f['value'] for f in flows]
)
)])
fig2.update_layout(
title_text="Customer Journey Sankey Diagram<br>Device → Browser → Purchase Outcome",
font=dict(size=12),
width=1000,
height=600
)
fig2.show()
print("\n** Sankey diagram created! **")
print(" -> Width of flows represents number of customers")
print(f"\n** Journey Summary: **")
print(f"Total Customers: {len(df_journey)}")
purchase_rate = ((df_journey['Outcome'] == 'Purchased').sum() / len(df_journey) * 100)
print(f"Purchase Rate: {purchase_rate:.1f}%")
print(f"Most Common Device: {df_journey['Device'].mode()[0]}")
print(f"Most Common Browser: {df_journey['Browser'].mode()[0]}")
```
