```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

```python


celsius = np.array([
    -273.15, -100, -60, -40, -30, -20, -15, -10, -5,   # extreme cold / subzero
      0,   1,   2,   3,   4,   5,  10,  12,  15,  18,  # freezing range
     20,  21,  22,  23,  24,  25,  27,  30,  33,  35,  # room temp range
     37, 37.5, 38, 38.5, 39,  40,  41,  42,  45,  50,  # body temp / fever
     55,  60,  65,  70,  75,  80,  85,  90,  95, 100,  # hot range
    110, 120, 150, 180, 200, 250, 300, 500, 1000, 5505  # industrial / extreme
], dtype=float)

fahrenheit = np.round(celsius * 9/5 + 32, 2).astype(float)
```

```python
#the first layer that takes inputs in this case we have a unique neuron
capa = tf.keras.layers.Dense(units=1,input_shape=[1]) #Dense layers they have conecction
#connection with all layers

model=tf.keras.Sequential([capa]) #sequential model basic model to work with

```

```python
"""
    adamś algortihm should be fittet with correct values
    if numbers are to great they will be to general, less value slow learning

    loss function = few great errors are worst
"""

model.compile(
    #not to overfit nor underfit
    optimizer=tf.keras.optimizers.Adam(0.1), #algortihm to adjust bias and values
    loss='mean_squared_error' #lost function
)

```

```python
"""
we should start training the model
fit function (input, output, epochs) how many epochs to train the model
"""

story = model.fit(celsius,fahrenheit,epochs=1000,verbose=False)
```

by understanding the curve we can anayze that we do not need more epoch with 200 was enough

```python
"""
analyzing loss function
with every epoch the loss function decreases

"""
plt.xlabel("# epoch")
plt.ylabel("loss")
plt.plot(story.history['loss'])
plt.show()
```

```python
cel = float(input("Enter a Celsius value: "))
result = model.predict(np.array([[cel]]))
print(f"{cel} celsius is {result[0][0]:.2f} fahrenheit")
```
